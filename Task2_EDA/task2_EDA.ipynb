## Exploratory Data Analysis

```python
import pandas as pd
import numpy as np

df = pd.read_csv('heart.csv')
```

```python
df.rename(columns = {
    'age' : 'Age',
    'sex' : 'Sex',
    'cp' : 'ChestPainType',
    'trestbps' : 'RestingBP',
    'chol' : 'Cholesterol',
    'fbs' : 'Fasting Blood Sugar',
    'restecg' : 'Resting ECG',
    'thalach' : 'Max Heart Rate',
    'exang' : 'Exercise Induced Angina',
    'oldpeak' : 'ST Depression',
    'slope' : 'Slope Of ST Segment',
    'ca' : 'Major Vessels Count',
    'thal' : 'Thalassemia',
    'target' : 'Heart Disease'
}, inplace = True)
df.columns.to_list()
```

```python
gender_mapping = {0: 'Female', 1: 'Male'}
df['Sex'] = df['Sex'].map(gender_mapping)

cp_mapping = {0: "Typical Angina", 1: "Atypical Angina", 2: "Non-Anginal Pain", 3: "Asymptomatic"}
df['ChestPainType'] = df['ChestPainType'].map(cp_mapping)

heart_mapping = {1: 'Has Heart Disease', 0: 'No Heart Disease'}
df['Heart Disease'] = df['Heart Disease'].map(heart_mapping)

fbs_mapping = {1: 'High (>120mg/dl)', 0: 'Normal'}
df['Fasting Blood Sugar'] = df['Fasting Blood Sugar'].map(fbs_mapping)

exang_mapping = {1: 'Yes', 0: 'No'}
df['Exercise Induced Angina'] = df['Exercise Induced Angina'].map(exang_mapping)

print(df.head(5))
```

## 2. Statistical summary
Key observations:

## 3. Research Questions
Before diving into analysis, we define the key questions we want to answer:

1. Does age increase the risk of heart disease?
2. Are males or females more at risk?
3. Does high cholesterol contribute to heart disease?
4. Does chest pain type predict heart disease?
5. Does maximum heart rate differ between patients with and without heart disease?
6. Which factors are most strongly correlated with heart disease?

```python
age_analysis = df.groupby('Heart Disease')['Age'].mean()
print("Average age by heart disease status:")
print(f"No Heart Disease (0): {age_analysis['No Heart Disease']:.1f} years")
print(f"Has Heart Disease (1): {age_analysis['Has Heart Disease']:.1f} years")
```

```python
gender_analysis = df.groupby('Heart Disease')['Sex'].value_counts()
print("Heart Disease by Gender:")
print(gender_analysis)
```

## 4. Key Findings

### Age Analysis
- Patients with heart disease are slightly younger (52.5 yrs) than those without (56.6 yrs)
- Suggests heart disease affects adults across all age groups, not just the elderly

### Gender Analysis
- 75% of female patients have heart disease vs 45% of male patients
- Women appear at higher risk in this dataset

```python
chol_analysis = df.groupby('Heart Disease')['Cholesterol'].mean()
print("Average Cholesterol by heart disease status:")
print(f"No Heart Disease : {chol_analysis['No Heart Disease']:.1f} mg/dl")
print(f"Has Heart Disease : {chol_analysis['Has Heart Disease']:.1f} mg/dl")
```

```python
chest_pain_analysis = df.groupby('ChestPainType')['Heart Disease'].value_counts()
print("Heart Disease by Chest Pain Type:")
print(chest_pain_analysis)
```

```python
max_heart_rate_analysis = df.groupby('Heart Disease')['Max Heart Rate'].mean()
print("Heart Disease by Max Heart Rate:")
print(f"No Heart Disease : {max_heart_rate_analysis['No Heart Disease']:.1f} bpm")
print(f"Has Heart Disease : {max_heart_rate_analysis['Has Heart Disease']:.1f} bpm")
```

## 4. Key Findings (continued)

### Cholesterol
- Patients WITHOUT heart disease have slightly higher cholesterol (251 vs 242 mg/dl)
- Cholesterol alone is not a strong predictor of heart disease in this dataset
- Other factors appear more significant

### Chest Pain Type
- Surprisingly, most Typical Angina patients do NOT have heart disease
- Asymptomatic patients are more likely to have heart disease
- Silent heart disease is more dangerous — no symptoms doesn't mean safe!

### Max Heart Rate
- Patients with heart disease achieve higher max heart rates (158 vs 139 bpm)
- Suggests the heart compensates by beating faster in affected patients

```python
df['Heart Disease Numeric'] = df['Heart Disease'].map({'Has Heart Disease': 1, 'No Heart Disease': 0})

correlation = df[['Age', 'Cholesterol', 'Max Heart Rate', 
                   'RestingBP', 'ST Depression',
                   'Heart Disease Numeric']].corr()

print("Correlation with Heart Disease:")
print(correlation['Heart Disease Numeric'].sort_values(ascending=False).round(3))
```

## 5. Correlation Analysis

Strongest predictors of heart disease:
- ST Depression (-0.431): strongest predictor overall
- Max Heart Rate (+0.422): higher rates associated with heart disease
- Age (-0.225): younger patients more affected
- Cholesterol (-0.085): surprisingly weak predictor

Note: Negative correlation means lower values are associated 
with presence of heart disease

```python
# Outlier Detection using IQR method
Q1 = df[['Age', 'Cholesterol', 'Max Heart Rate', 'RestingBP', 'ST Depression']].quantile(0.25)
Q3 = df[['Age', 'Cholesterol', 'Max Heart Rate', 'RestingBP', 'ST Depression']].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df[['Age', 'Cholesterol', 'Max Heart Rate', 'RestingBP', 'ST Depression']] < (Q1 - 1.5 * IQR)) | 
            (df[['Age', 'Cholesterol', 'Max Heart Rate', 'RestingBP', 'ST Depression']] > (Q3 + 1.5 * IQR))).sum()

print("Number of Outliers per Column:")
print(outliers)
```

## 6. Outlier Detection

Using the IQR (Interquartile Range) method to detect outliers:
